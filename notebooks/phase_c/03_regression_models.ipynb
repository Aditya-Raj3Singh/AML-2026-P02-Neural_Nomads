# Phase C — Regression Models for Next-Day PM2.5 Forecasting

**City:** Delhi

## Objective

This notebook trains and evaluates regression models for predicting next-day
PM2.5 concentration.

The following models are evaluated:

1. Linear Regression
2. Ridge Regression
3. Lasso Regression

The evaluation metrics are:

- MAE
- RMSE
- R²

A time-ordered train-test split is used.

Ridge and Lasso hyperparameters are selected using time-series
cross-validation.

All models are compared with the persistence baseline.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import (
    LinearRegression,
    RidgeCV,
    LassoCV
)

from sklearn.model_selection import TimeSeriesSplit

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import warnings
warnings.filterwarnings("ignore")

print("Libraries imported successfully.")
```

```python
INPUT_FILE = "delhi_regression_features.csv"

df = pd.read_csv(INPUT_FILE)

df["Date"] = pd.to_datetime(df["Date"])

df = df.sort_values("Date").reset_index(drop=True)

print("Dataset loaded successfully.")
print("Shape:", df.shape)

print(
    "Date range:",
    df["Date"].min(),
    "to",
    df["Date"].max()
)

print("\nColumns:")
print(df.columns.tolist())

df.head()
```

```python
FEATURES = [
    "PM2.5_current",
    "PM2.5_lag1",
    "PM2.5_lag2",
    "PM2.5_lag3",
    "PM2.5_lag7",
    "PM2.5_lag14",
    "PM2.5_roll3_mean",
    "PM2.5_roll7_mean",
    "PM2.5_roll14_mean",
    "PM2.5_roll7_std",
    "month",
    "day_of_week",
    "day_of_year"
]

TARGET = "PM2.5_tomorrow"

required_columns = (
    ["Date"] +
    FEATURES +
    [TARGET]
)

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        "Wrong input file. Missing columns: "
        + str(missing_columns)
    )

print("Correct engineered dataset loaded.")
print("All required columns found.")
```

```python
X = df[FEATURES].copy()
y = df[TARGET].copy()

print("X shape:", X.shape)
print("y shape:", y.shape)

print(
    "Missing predictor values:",
    X.isna().sum().sum()
)

print(
    "Missing target values:",
    y.isna().sum()
)
```

## Time-Ordered Train-Test Split

Data through 31 December 2019 is used for training.

Data from 2020 is reserved for testing.

Random splitting is avoided because it would mix past and future
observations.

The same test dates are used for the persistence baseline and all regression
models.

```python
TRAIN_END = pd.Timestamp("2019-12-31")

train_mask = df["Date"] <= TRAIN_END
test_mask = df["Date"] > TRAIN_END

X_train = X.loc[train_mask].copy()
X_test = X.loc[test_mask].copy()

y_train = y.loc[train_mask].copy()
y_test = y.loc[test_mask].copy()

dates_train = df.loc[
    train_mask,
    "Date"
].copy()

dates_test = df.loc[
    test_mask,
    "Date"
].copy()

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

print(
    "\nTraining:",
    dates_train.min().date(),
    "to",
    dates_train.max().date()
)

print(
    "Testing:",
    dates_test.min().date(),
    "to",
    dates_test.max().date()
)
```

```python
assert len(X_train) > 0
assert len(X_test) > 0

assert dates_train.max() < dates_test.min()

assert y_train.isna().sum() == 0
assert y_test.isna().sum() == 0

print("Time-ordered split verification PASSED.")
```

```python
def evaluate_model(name, y_true, y_pred):

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )

    r2 = r2_score(
        y_true,
        y_pred
    )

    return {
        "Model": name,
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2
    }
```

## Persistence Baseline

The persistence model assumes:

**Tomorrow's PM2.5 = Today's PM2.5**

This is the benchmark against which the regression models are evaluated.

```python
persistence_pred = X_test["PM2.5_current"].copy()

valid_persistence = (
    persistence_pred.notna()
    & y_test.notna()
)

persistence_results = evaluate_model(
    "Persistence Baseline",
    y_test.loc[valid_persistence],
    persistence_pred.loc[valid_persistence]
)

print("Persistence Baseline Results")
print(persistence_results)
```

## Linear Regression

```python
linear_model = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    ),
    (
        "scaler",
        StandardScaler()
    ),
    (
        "model",
        LinearRegression()
    )
])

linear_model.fit(
    X_train,
    y_train
)

linear_pred = linear_model.predict(
    X_test
)

linear_results = evaluate_model(
    "Linear Regression",
    y_test,
    linear_pred
)

print("Linear Regression Results")
print(linear_results)
```

## Time-Series Cross-Validation

```python
tscv = TimeSeriesSplit(
    n_splits=5
)

print(
    "TimeSeriesSplit folds:",
    tscv.n_splits
)
```

## Ridge Regression

```python
ridge_alphas = np.logspace(
    -3,
    3,
    50
)

ridge_model = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    ),
    (
        "scaler",
        StandardScaler()
    ),
    (
        "model",
        RidgeCV(
            alphas=ridge_alphas,
            cv=tscv,
            scoring="neg_mean_absolute_error"
        )
    )
])

ridge_model.fit(
    X_train,
    y_train
)

ridge_pred = ridge_model.predict(
    X_test
)

ridge_results = evaluate_model(
    "Ridge Regression",
    y_test,
    ridge_pred
)

print("Ridge Regression Results")
print(ridge_results)
```

```python
best_ridge_alpha = (
    ridge_model
    .named_steps["model"]
    .alpha_
)

print(
    "Best Ridge alpha:",
    best_ridge_alpha
)
```

## Lasso Regression

```python
lasso_model = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    ),
    (
        "scaler",
        StandardScaler()
    ),
    (
        "model",
        LassoCV(
            alphas=np.logspace(
                -3,
                2,
                100
            ),
            cv=tscv,
            max_iter=50000
        )
    )
])

lasso_model.fit(
    X_train,
    y_train
)

lasso_pred = lasso_model.predict(
    X_test
)

lasso_results = evaluate_model(
    "Lasso Regression",
    y_test,
    lasso_pred
)

print("Lasso Regression Results")
print(lasso_results)
```

```python
best_lasso_alpha = (
    lasso_model
    .named_steps["model"]
    .alpha_
)

print(
    "Best Lasso alpha:",
    best_lasso_alpha
)
```

## Final Comparison

```python
results = pd.DataFrame([
    persistence_results,
    linear_results,
    ridge_results,
    lasso_results
])

results = (
    results
    .sort_values("RMSE")
    .reset_index(drop=True)
)

print("FINAL MODEL COMPARISON")

display(
    results.round(3)
)
```

## Improvement vs Persistence

```python
baseline_mae = persistence_results["MAE"]
baseline_rmse = persistence_results["RMSE"]

comparison = results.copy()

comparison["MAE_Improvement_%"] = (
    (
        baseline_mae
        - comparison["MAE"]
    )
    / baseline_mae
    * 100
)

comparison["RMSE_Improvement_%"] = (
    (
        baseline_rmse
        - comparison["RMSE"]
    )
    / baseline_rmse
    * 100
)

display(
    comparison.round(2)
)
```

## Best Model

```python
regression_results = results[
    results["Model"] != "Persistence Baseline"
].copy()

best_row = (
    regression_results
    .sort_values("RMSE")
    .iloc[0]
)

best_model_name = best_row["Model"]

print(
    "Best regression model:",
    best_model_name
)

print(
    "MAE:",
    round(best_row["MAE"], 3)
)

print(
    "RMSE:",
    round(best_row["RMSE"], 3)
)

print(
    "R²:",
    round(best_row["R2"], 3)
)
```

## Coefficient Comparison

```python
coefficient_table = pd.DataFrame({
    "Feature": FEATURES,

    "Linear":
        linear_model
        .named_steps["model"]
        .coef_,

    "Ridge":
        ridge_model
        .named_steps["model"]
        .coef_,

    "Lasso":
        lasso_model
        .named_steps["model"]
        .coef_
})

coefficient_table["Ridge_Abs"] = (
    coefficient_table["Ridge"].abs()
)

coefficient_table = (
    coefficient_table
    .sort_values(
        "Ridge_Abs",
        ascending=False
    )
)

display(coefficient_table)
```
